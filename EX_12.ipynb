```python
from collections import deque


def water_jug_problem(x, y, z):
    """
    Solves the Water Jug Problem using Breadth-First Search.

    x: capacity of Jug 1
    y: capacity of Jug 2
    z: target amount of water
    """

    queue = deque([(0, 0, [])])

    # Start with both jugs empty
    visited = set()

    while queue:
        a, b, steps = queue.popleft()

        if (a, b) in visited:
            continue

        visited.add((a, b))

        # Check whether target is reached
        if a == z or b == z:
            return steps

        # Fill Jug 1
        queue.append(
            (x, b, steps + ['Fill Jug 1'])
        )

        # Fill Jug 2
        queue.append(
            (a, y, steps + ['Fill Jug 2'])
        )

        # Empty Jug 1
        queue.append(
            (0, b, steps + ['Empty Jug 1'])
        )

        # Empty Jug 2
        queue.append(
            (a, 0, steps + ['Empty Jug 2'])
        )

        # Pour Jug 1 into Jug 2
        amount = min(a, y - b)

        queue.append(
            (a - amount,
             b + amount,
             steps + ['Pour Jug 1 into Jug 2'])
        )

        # Pour Jug 2 into Jug 1
        amount = min(x - a, b)

        queue.append(
            (a + amount,
             b - amount,
             steps + ['Pour Jug 2 into Jug 1'])
        )

    return None


# Jug capacities: 4 litres and 3 litres
# Target: 2 litres
steps = water_jug_problem(4, 3, 2)

if steps:
    print("\n".join(steps))
else:
    print("No solution found.")
```
