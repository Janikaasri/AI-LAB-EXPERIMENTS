```python
def alphabeta(depth, node, alpha, beta, maximizing):
    if depth == 0:
        return node

    if maximizing:
        value = float('-inf')

        for x in [node - 1, node + 1]:
            value = max(value, alphabeta(depth - 1, x, alpha, beta, False))
            alpha = max(alpha, value)

            if alpha >= beta:
                break

        return value

    else:
        value = float('inf')

        for x in [node - 1, node + 1]:
            value = min(value, alphabeta(depth - 1, x, alpha, beta, True))
            beta = min(beta, value)

            if alpha >= beta:
                break

        return value


print("Alpha-Beta Value:",
      alphabeta(3, 5, float('-inf'), float('inf'), True))
```
