```python
students = {
    "Ravi": ("Anitha", "CS101"),
    "Priya": ("Kumar", "CS102"),
    "Arun": ("Anitha", "CS101")
}

name = input("Enter student name: ")

if name in students:
    teacher, code = students[name]
    print("Teacher:", teacher)
    print("Subject Code:", code)
else:
    print("Student not found")
```
