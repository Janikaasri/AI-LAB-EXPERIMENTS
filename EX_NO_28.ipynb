```python
planets = {
    "Mercury": "Smallest planet",
    "Venus": "Hottest planet",
    "Earth": "Planet with life",
    "Mars": "Red planet",
    "Jupiter": "Largest planet",
    "Saturn": "Planet with rings",
    "Uranus": "Ice giant",
    "Neptune": "Farthest planet"
}

name = input("Enter planet name: ")

if name in planets:
    print(planets[name])
else:
    print("Planet not found")
```
