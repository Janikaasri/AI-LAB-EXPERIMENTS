```python
facts = {"rain"}

rules = {
    "rain": "wet_ground",
    "wet_ground": "slippery"
}

changed = True

while changed:
    changed = False

    for condition, result in rules.items():
        if condition in facts and result not in facts:
            facts.add(result)
            changed = True

print("Facts:", facts)
```
