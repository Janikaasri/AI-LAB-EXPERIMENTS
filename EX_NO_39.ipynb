```python
facts = {"rain"}

rules = {
    "wet_ground": "rain",
    "slippery": "wet_ground"
}


def backward_chaining(goal):
    if goal in facts:
        return True

    if goal in rules:
        condition = rules[goal]

        if backward_chaining(condition):
            return True

    return False


goal = input("Enter goal: ")

if backward_chaining(goal):
    print("Goal can be proved.")
else:
    print("GoRAal cannot be proved.")
```
