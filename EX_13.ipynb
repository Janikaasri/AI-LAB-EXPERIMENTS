```python
from itertools import permutations


def solve_cryptarithmetic(puzzle):
    """
    Solves a cryptarithmetic puzzle.
    Example: SEND + MORE = MONEY
    """

    # Split the puzzle
    parts = puzzle.replace("=", "+").split("+")

    word1 = parts[0].strip()
    word2 = parts[1].strip()
    result = parts[2].strip()

    # Find all unique letters
    letters = set(word1 + word2 + result)

    # More than 10 letters cannot be assigned digits
    if len(letters) > 10:
        return None

    letters = list(letters)

    # Check every possible digit assignment
    for perm in permutations(range(10), len(letters)):
        mapping = dict(zip(letters, perm))

        # Leading letters cannot be zero
        if mapping[word1[0]] == 0:
            continue

        if mapping[word2[0]] == 0:
            continue

        if mapping[result[0]] == 0:
            continue

        # Convert words into numbers
        num1 = int("".join(str(mapping[c]) for c in word1))
        num2 = int("".join(str(mapping[c]) for c in word2))
        num_result = int("".join(str(mapping[c]) for c in result))

        # Check the equation
        if num1 + num2 == num_result:
            return mapping

    return None


# Example usage
puzzle = "SEND + MORE = MONEY"

mapping = solve_cryptarithmetic(puzzle)

if mapping:
    print("Solution:")
    print(mapping)

    send = "".join(str(mapping[c]) for c in "SEND")
    more = "".join(str(mapping[c]) for c in "MORE")
    money = "".join(str(mapping[c]) for c in "MONEY")

    print(send, "+", more, "=", money)

else:
    print("No solution found.")
```
