```python
symptoms = input("Enter symptoms: ").lower()

if "fever" in symptoms and "cough" in symptoms:
    print("Diagnosis: Flu")

elif "fever" in symptoms and "body pain" in symptoms:
    print("Diagnosis: Viral Fever")

elif "headache" in symptoms and "cold" in symptoms:
    print("Diagnosis: Common Cold")

else:
    print("Diagnosis: Unknown")
```
