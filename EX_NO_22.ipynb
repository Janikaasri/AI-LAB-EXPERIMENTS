```python
def minimax(depth, node, maximizing):
    if depth == 0:
        return node

    if maximizing:
        return max(
            minimax(depth - 1, node - 1, False),
            minimax(depth - 1, node + 1, False)
        )
    else:
        return min(
            minimax(depth - 1, node - 1, True),
            minimax(depth - 1, node + 1, True)
        )


print("Minimax Value:", minimax(3, 5, True))
```
