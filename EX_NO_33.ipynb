```python
diet = {
    "diabetes": "Eat vegetables, whole grains and avoid sugar.",
    "obesity": "Eat fruits, vegetables and low-calorie food.",
    "anemia": "Eat spinach, beans and iron-rich foods.",
    "hypertension": "Eat low-salt and potassium-rich foods."
}

disease = input("Enter disease: ").lower()

if disease in diet:
    print("Diet:", diet[disease])
else:
    print("Disease not found")
```
