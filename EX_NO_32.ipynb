```python
female = ["Pam", "Liz", "Ann", "Pat"]
male = ["Tom", "Bob", "Jim"]

mother = {
    "Liz": "Pam",
    "Ann": "Pam"
}

father = {
    "Liz": "Tom",
    "Ann": "Tom"
}

def check_mother(child):
    if child in mother:
        print("Mother:", mother[child])
    else:
        print("Mother not found")


def check_father(child):
    if child in father:
        print("Father:", father[child])
    else:
        print("Father not found")


name = input("Enter child name: ")

check_mother(name)
check_father(name)
```
